```python
# function to calculate business day between 2 dates
def business_days_without_holidays(start, end, hollyday):
    days = pd.date_range(start, end, freq="D")
    return sum(
        1 for d in days
        if d.weekday() < 5 and d.date() not in hollyday
    )
```

```python
def business_hour_without_holiday(start, end, hollydays):
    if pd.isna(start) or pd.isna(end):
        return None
    
    # gera intervalo por hora
    hours = pd.date_range(start, end, freq="h")
    
    # filtra apenas dias úteis e remove feriados
    valid_hours = [
        h for h in hours
        if h.weekday() < 5 and h.date() not in hollydays
    ]
    
    return len(valid_hours)
```
